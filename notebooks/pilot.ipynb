```python
import numpy as np, torch, torch.nn as nn, torch.nn.functional as F, copy, time

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

npz = np.load("eeg_data.npz")
X = {s: npz[f"X{s}"] for s in range(1, 10)}
y = {s: npz[f"y{s}"] for s in range(1, 10)}

class EEGNet(nn.Module):
    def __init__(self, n_chans=22, n_classes=4, F1=8, D=2, F2=16, drop=0.5):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, F1, (1, 64), padding=(0, 32), bias=False),
            nn.BatchNorm2d(F1),
            nn.Conv2d(F1, F1 * D, (n_chans, 1), groups=F1, bias=False),
            nn.BatchNorm2d(F1 * D),
            nn.ELU(),
            nn.AvgPool2d((1, 4)),
            nn.Dropout(drop),
            nn.Conv2d(F1 * D, F1 * D, (1, 16), padding=(0, 8), groups=F1 * D, bias=False),
            nn.Conv2d(F1 * D, F2, 1, bias=False),
            nn.BatchNorm2d(F2),
            nn.ELU(),
            nn.AvgPool2d((1, 8)),
            nn.Dropout(drop),
            nn.Flatten(),
        )
        self.classifier = nn.LazyLinear(n_classes)

    def forward(self, x):
        return self.classifier(self.features(x))

def train_model(train_subjects, epochs=30, seed=0):
    torch.manual_seed(seed)
    np.random.seed(seed)
    Xtr = np.concatenate([X[s] for s in train_subjects])
    ytr = np.concatenate([y[s] for s in train_subjects])
    scale = float(Xtr.std())
    Xtr = torch.tensor(Xtr / scale).unsqueeze(1).to(device)
    ytr = torch.tensor(ytr, dtype=torch.long).to(device)
    model = EEGNet().to(device)
    model.eval()
    with torch.no_grad():
        model(Xtr[:2])  # initialize the lazy linear layer
    opt = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)
    n = len(Xtr)
    for ep in range(epochs):
        model.train()
        perm = torch.randperm(n, device=device)
        for i in range(0, n, 64):
            idx = perm[i:i + 64]
            loss = F.cross_entropy(model(Xtr[idx]), ytr[idx])
            opt.zero_grad()
            loss.backward()
            opt.step()
    return model, scale

@torch.no_grad()
def evaluate(model, Xte, yte, scale):
    model.eval()
    Xt = torch.tensor(Xte / scale).unsqueeze(1).to(device)
    pred = torch.cat([model(Xt[i:i + 256]).argmax(1).cpu() for i in range(0, len(Xt), 256)])
    return float((pred.numpy() == yte).mean())

results, models = {}, {}
for test_s in range(1, 10):
    train_s = [s for s in range(1, 10) if s != test_s]
    t0 = time.time()
    model, scale = train_model(train_s)
    acc = evaluate(model, X[test_s], y[test_s], scale)
    results[test_s] = acc
    models[test_s] = (copy.deepcopy(model.state_dict()), scale)
    print(f"subject {test_s}: acc={acc:.3f} ({time.time() - t0:.0f}s)")

print("mean acc:", np.mean(list(results.values())))
torch.save(models, "baseline_models.pt")
np.save("baseline_results.npy", results)
```

```python
from scipy.linalg import sqrtm, inv

def ea(Xs):
    R = np.mean([x @ x.T for x in Xs], axis=0)
    W = inv(sqrtm(R)).real.astype(np.float32)
    return np.stack([W @ x for x in Xs])

X_ea = {s: ea(X[s]) for s in range(1, 10)}
print("EA data std:", X_ea[1].std())

def train_model(train_subjects, epochs=30, seed=0, data_X=None):
    data_X = X if data_X is None else data_X
    torch.manual_seed(seed)
    np.random.seed(seed)
    Xtr = np.concatenate([data_X[s] for s in train_subjects])
    ytr = np.concatenate([y[s] for s in train_subjects])
    scale = float(Xtr.std())
    Xtr = torch.tensor(Xtr / scale).unsqueeze(1).to(device)
    ytr = torch.tensor(ytr, dtype=torch.long).to(device)
    model = EEGNet().to(device)
    model.eval()
    with torch.no_grad():
        model(Xtr[:2])
    opt = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)
    n = len(Xtr)
    for ep in range(epochs):
        model.train()
        perm = torch.randperm(n, device=device)
        for i in range(0, n, 64):
            idx = perm[i:i + 64]
            loss = F.cross_entropy(model(Xtr[idx]), ytr[idx])
            opt.zero_grad()
            loss.backward()
            opt.step()
    return model, scale

baseline = np.load("baseline_results.npy", allow_pickle=True).item()
results_ea, models_ea = {}, {}
for test_s in range(1, 10):
    train_s = [s for s in range(1, 10) if s != test_s]
    model, scale = train_model(train_s, data_X=X_ea)
    acc = evaluate(model, X_ea[test_s], y[test_s], scale)
    results_ea[test_s] = acc
    models_ea[test_s] = (copy.deepcopy(model.state_dict()), scale)
    print(f"subject {test_s}: baseline={baseline[test_s]:.3f}  EA={acc:.3f}  diff={acc - baseline[test_s]:+.3f}")

worse = sum(results_ea[s] < baseline[s] for s in range(1, 10))
print("mean baseline:", np.mean(list(baseline.values())))
print("mean EA:", np.mean(list(results_ea.values())))
print("subjects worse than baseline:", worse, "of 9")
torch.save(models_ea, "ea_models.pt")
np.save("ea_results.npy", results_ea)
```

```python
import math

def load_model(state):
    m = EEGNet().to(device)
    m.eval()
    with torch.no_grad():
        m(torch.zeros(2, 1, 22, 1001, device=device))
    m.load_state_dict(state)
    return m

def tent_acc(state, scale, Xte, yte, lr=1e-3, bs=32, ent_thresh=None):
    m = load_model(state)
    for p in m.parameters():
        p.requires_grad_(False)
    m.eval()  # dropout off
    bn_params = []
    for mod in m.modules():
        if isinstance(mod, nn.BatchNorm2d):
            mod.train()  # use batch statistics
            mod.weight.requires_grad_(True)
            mod.bias.requires_grad_(True)
            bn_params += [mod.weight, mod.bias]
    opt = torch.optim.Adam(bn_params, lr=lr) if lr > 0 else None
    Xt = torch.tensor(Xte / scale).unsqueeze(1).to(device)
    preds = []
    for i in range(0, len(Xt), bs):
        logits = m(Xt[i:i + bs])
        preds.append(logits.argmax(1).detach().cpu())
        if opt is None:
            continue
        p = logits.softmax(1)
        ent = -(p * p.clamp_min(1e-8).log()).sum(1)
        if ent_thresh is None:
            keep = torch.ones_like(ent, dtype=torch.bool)
        else:
            keep = ent < ent_thresh
        if keep.any():
            opt.zero_grad()
            ent[keep].mean().backward()
            opt.step()
    return float((torch.cat(preds).numpy() == yte).mean())

thr = 0.4 * math.log(4)
rows = {}
for s in range(1, 10):
    st, sc = models[s]
    st_ea, sc_ea = models_ea[s]
    rows[s] = dict(
        base=baseline[s],
        ea=results_ea[s],
        bn_only=tent_acc(st, sc, X[s], y[s], lr=0),
        tent=tent_acc(st, sc, X[s], y[s]),
        ea_tent=tent_acc(st_ea, sc_ea, X_ea[s], y[s]),
        ea_tent_filter=tent_acc(st_ea, sc_ea, X_ea[s], y[s], ent_thresh=thr),
    )
    print(s, {k: round(v, 3) for k, v in rows[s].items()})

print("--- summary ---")
for k in ["base", "ea", "bn_only", "tent", "ea_tent", "ea_tent_filter"]:
    vals = [rows[s][k] for s in range(1, 10)]
    worse_base = sum(rows[s][k] < rows[s]["base"] for s in range(1, 10))
    worse_ea = sum(rows[s][k] < rows[s]["ea"] for s in range(1, 10))
    print(f"{k:16s} mean={np.mean(vals):.3f}  worse_than_base={worse_base}  worse_than_ea={worse_ea}")
```

```python
all_runs = {0: {k: [rows[s][k] for s in range(1, 10)]
                for k in ["base", "ea", "bn_only", "tent", "ea_tent", "ea_tent_filter"]}}

for seed in [1, 2]:
    base_states, ea_states = {}, {}
    for s in range(1, 10):
        tr = [t for t in range(1, 10) if t != s]
        m, sc = train_model(tr, seed=seed)
        base_states[s] = (copy.deepcopy(m.state_dict()), sc)
        m2, sc2 = train_model(tr, seed=seed, data_X=X_ea)
        ea_states[s] = (copy.deepcopy(m2.state_dict()), sc2)
    run = {k: [] for k in all_runs[0]}
    for s in range(1, 10):
        st, sc = base_states[s]
        st2, sc2 = ea_states[s]
        run["base"].append(evaluate(load_model(st), X[s], y[s], sc))
        run["ea"].append(evaluate(load_model(st2), X_ea[s], y[s], sc2))
        run["bn_only"].append(tent_acc(st, sc, X[s], y[s], lr=0))
        run["tent"].append(tent_acc(st, sc, X[s], y[s]))
        run["ea_tent"].append(tent_acc(st2, sc2, X_ea[s], y[s]))
        run["ea_tent_filter"].append(tent_acc(st2, sc2, X_ea[s], y[s], ent_thresh=thr))
    all_runs[seed] = run
    print("seed", seed, {k: round(float(np.mean(v)), 3) for k, v in run.items()})

print("--- mean +- std over seeds ---")
for k in all_runs[0]:
    means = [np.mean(all_runs[sd][k]) for sd in all_runs]
    print(f"{k:16s} {np.mean(means):.3f} +- {np.std(means):.3f}")
np.save("all_runs.npy", all_runs)
```

```python
import pandas as pd

records = []
for seed, run in all_runs.items():
    for method, accs in run.items():
        for subj, acc in zip(range(1, 10), accs):
            records.append(dict(seed=seed, method=method, subject=subj,
                                accuracy=round(float(acc), 4)))
df = pd.DataFrame(records)
df.to_csv("results.csv", index=False)

base = df[df.method == "base"].set_index(["seed", "subject"]).accuracy
for m in ["base", "ea", "bn_only", "tent", "ea_tent", "ea_tent_filter"]:
    sub = df[df.method == m].set_index(["seed", "subject"]).accuracy
    seed_means = sub.groupby("seed").mean()
    worse = (sub < base).groupby("seed").sum()
    print(f"{m:16s} mean={seed_means.mean()*100:.1f} std={seed_means.std(ddof=0)*100:.1f} "
          f"worse_than_base_per_seed={worse.tolist()}")
```
